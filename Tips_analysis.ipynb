## Tips Data Analysis.
El fin de este proyecto es analizar el comportamiento de las propinas en relación con el total de la cuenta, el número de personas y otras variables.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)
print(df.head())
```

## Datos a estudiar.
- total_bill: total de la cuenta.
- tip: propina.
- size: número de personas.


```python
df.describe()
```

## Análisis descriptivo.

El promedio de la cuenta total es de aproximadamente 19.8 dólares.

La desviación estándar es de aproximadamente 8.9, lo que indica que existe una cierta variabilidad en los montos de las cuentas.

La mayoría de las cuentas se encuentra en un rango aproximado entre 10.9 y 28.7 dólares.

Para las propinas, se tiene que el promedio es de aproximadamente 3 dolares y la desviación estándar es de 1.4, esto indica que las propinas son relativamente consistentes. 

Finalmente para el número de personas, tenemos que el promedio es de 2.5 personas por mesa, y dado que su desviación es de 1, tenemos que la mayoría las mesas tiene entre 2 y 3 personas.

Si analizamos los cuartiles podemos observar que el 75% de cuentas pagan menos de 24.13 dolares.

## Diagrama de dispersión.

```python
plt.scatter(df["total_bill"], df["tip"])
plt.xlabel("Total de la cuenta")
plt.ylabel("Propina")
plt.title("Relación entre cuenta y propina")
plt.show()
```

## Relación entre el total de la cuenta y propina.

De la gráfica podemos observar que hay una correlación positiva entre el total de la cuenta y la propina.
Esto indica que, en general, a mayor consumo, mayor propina.

Sin embargo, la relación no es perfecta, pues existen cuentas altas con propinas bajas.

## Modelo de regresión.

La recta de regresión es de la forma:

y=mx+b

Donde:
- y: propina.
- x: total de la cuenta.

```python
X = df[["total_bill"]]
y = df["tip"]
modelo = LinearRegression()
modelo.fit(X, y)
print("Pendiente:", modelo.coef_[0])
print("Intercepto:", modelo.intercept_)
```

El intercepto es aproximadamente de 0.92, lo que puede interpretarse como una propina base.

La pendiente es aproximadamente de 0.105, lo que sugiere que en promedio los clientes dejan alrededor del 10.5% del total de la cuenta como propina.

## Conclusion.

Existe una relación positiva entre el total de la cuenta y la propina, aunque no es perfectamente lineal. 

La mayoría de las propinas se situán entre 2 y 4 dólares, y las cuentas más comunes están entre 10 y 20 dólares.

En cuentas altas se observa una mayor variabilidad, lo que sugiere que el comportamiento es menor predecible en estos casos.

El modelo de regresión indica que, en promedio, los clientes dejan alrededor del 10.5% como propina.
